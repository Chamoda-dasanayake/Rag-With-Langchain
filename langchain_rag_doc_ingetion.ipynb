```python
docs_dir_path = "C:/Users/HP/Desktop/Y3 S2/projects/RagWithLangchain/docs_dir"
vector_db_path = "C:/Users/HP/Desktop/Y3 S2/projects/RagWithLangchain/vector_db"
collection_name = "document_collection"
```

```python
embedding = HuggingFaceEmbeddings()
```

```python
# directory loaded
loader = DirectoryLoader(
    path=docs_dir_path,
    glob="./*.pdf",
    loader_cls=UnstructuredFileLoader
)
# single file can also be loaded with UnstructuredFileLoader
```

```python
# load the documents
documents = loader.load()
```

```python
# initializing the text splitters 
text_splitter = CharacterTextSplitter(
    chunk_size=2000,
    chunk_overlap=500
)
```

```python
# splitting the text into smallre chunks
text_chunks= text_splitter.split_documents(documents)
```

```python
# creating the vector store
vector_store = Chroma(
    persist_directory=vector_db_path,
    embedding_function=embedding,
    collection_name=collection_name
)
```
